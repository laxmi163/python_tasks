```python
#polymorphism
class Address:
  def __init__(self,street,city,state,zip):
     self.street=street
     self.city=city
     self.state=state
     self.zip=zip
  def add(self):
      print(self.street)
      print(self.city)
      print(self.state)
      print(self.zip)
class Employee:
  def __init__(self,employee_id,employee_name,address):
     self.employee_id=employee_id
     self.employee_name=employee_name
     self.address=address
  def emp(self):
      print(self.employee_id)
      print(self.employee_name)
      self.address.add()

class JuniorEngineer(Employee):
  def __init__(self,employee_id,employee_name,assessement_score,feedback,address):
     super().__init__(employee_id,employee_name,address)
     self.assessement_score = assessement_score
     self.feedback = feedback
     #method overriding
  def display(self):
      self.emp()
      print(self.assessement_score)
      print(self.feedback)

class SoftwareEngineer(Employee):
  def __init__(self,employee_id,employee_name,project_name,address):
     super().__init__(employee_name,employee_id,address)
     self.project_name = project_name
     #method overriding
  def display(self):
      self.emp()
      print(self.project_name)
class Trainer(Employee):
  def __init__(self,employee_name,employee_id,skills,certifications,address):
      super().__init__(employee_name,employee_id,address)
      self.skills = skills
      self.certifications = certifications
      #method overridiing
  def display(self):
       self.emp()
       print(self.skills)
       print(self.certifications)
address= Address("vasu nagar","darsi","ap",523247)
employee=Employee(548,"laxmi",address)

junior = JuniorEngineer(101, "Lakshmi",95,"good",address)

print(junior.__dict__)
software = SoftwareEngineer(102, "Rahul", "AI Resume Builder",address)
print(software.__dict__)
trainer = Trainer(103, "Anu","Python", "AWS",address)
print(trainer.__dict__)
print("-------junioregineer------")
junior.display()
print("-----address------")
address.add()
print("-------software------")
software.display()
print("------trainer------")
trainer.display()




```
