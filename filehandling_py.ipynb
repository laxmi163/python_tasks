```python
file=open("attendance.txt","x")
```

```python

with open("attendance.txt","w") as file:
    file.write("""rahul,absent
anu,present
laxmi,present
ravi,absent
priya,present""")
```

```python
with open("attendance.txt","r") as file:
  records=file.readlines()
total_students=len(records)
present_students=0
absent_students=0
for record in records:
  name, status=record.strip().split(",")
  if status=="present":
     present_students+=1
  else:
     absent_students+=1
report=f""" attendance report
--------
total_students:{total_students}
present_students:{present_students}
absent_students:{absent_students}"""

```

```python
with open("report.txt","w") as file:
  file.write(report)
```

```python
with open("report.txt","r") as file:
  print(file.read())
```

```python
file=open("logs.txt","x")
```

```python
with open("logs.txt","w") as file:
  file.write("""INFO: User logged in
ERROR: Database connection failed
INFO: File uploaded
ERROR: Invalid password
INFO: User logged out""")
```

```python
error_count = 0

with open("logs.txt", "r") as log_file, open("errors.txt", "w") as error_file:

    for line in log_file:
        if "ERROR" in line:
            error_file.write(line)
            error_count += 1

print("Total Errors Found:", error_count)
```

```python
with open ("errors.txt","r") as file:
  print(file.read())
```
