```python
class DayCalculate:
  @staticmethod
  def find_day(day,k):
   days=["sunday","monday","tuesday","wednesday","thursday","friday","saturday"]
   index=days.index(day)
   new_index=(index+k)%7
   return days[new_index]
print(DayCalculate.find_day("sunday",3))
```

```python
class DuplicateCharacter:

    @staticmethod
    def find_duplicates(text):

        freq = {}

        for ch in text.lower():

            if ch != " ":
                freq[ch] = freq.get(ch, 0) + 1

        print("Duplicate Characters:")

        for key, value in freq.items():

            if value > 1:
                print(f"{key} -> {value}")


DuplicateCharacter.find_duplicates("ABC Company pvt lmt")
```
